### What is spaCy?

spaCy is an open-source software library for advanced Natural Language Processing (NLP) in Python. It's designed specifically for production use and helps you build applications that process and understand large volumes of text.

It comes with pre-trained statistical models and word vectors, and it supports deep learning workflows. Some common tasks spaCy can perform include:

*   **Tokenization**: Breaking text into words or sentences.
*   **Part-of-speech (POS) tagging**: Assigning word types (e.g., noun, verb, adjective).
*   **Named Entity Recognition (NER)**: Identifying 'real-world objects' like people, organizations, or locations.
*   **Dependency Parsing**: Analyzing the grammatical structure of sentences.
*   **Lemmatization**: Reducing words to their base form.

Now that spaCy and the English model are installed, let's load the model and process some text.

```python
import spacy
```

```python
processor=spacy.load("en_core_web_sm")
```

```python
doc=processor(sample_text)
```
