```python
import langchain_huggingface as hf
```

```python
huf = hf.llms.HuggingFacePipeline.from_model_id(
    model_id="gpt2",
    task="text-generation",
    pipeline_kwargs={"max_new_tokens": 100},
)
```

```python
from langchain_core.prompts import PromptTemplate
```

```python
prompt=""
```

```python
prompt_llm = PromptTemplate.from_template(prompt)
```

```python
llm = prompt_llm | huf
```

```python
question="what is an LLM"
```

```python
llm.invoke(question)
```

```python
def llm_response(message,history):
  return llm.invoke({"Question":message})
```

```python
import gradio as gr

gr.ChatInterface(
    fn=llm_response,
    type="messages"
).launch()
```
